```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    print("No API key was found - fix it!")
else:
    print("API key found - looks good so far!")
```

```python
# constants
MODEL_GEMINI = 'gemini-3.1-flash-lite'
MODEL_GEMA = 'gemma-4-e2b'
```

### Via Calling an API Model

```python
# set up environment
from openai import AsyncOpenAI

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = AsyncOpenAI(api_key=api_key, base_url=GEMINI_BASE_URL)
```

```python
# here is the question
question = """
    I am a junior project manager and I have been asked to manage a project whose scope is to revamp our company's loyal customer CRM.
    I need to create a project plan that outlines the key milestones, deliverables, and resources required for the successful completion of the project.
    Additionally, I need to identify potential risks and develop mitigation strategies to ensure that the project stays on track.
    Could you help me by providing comprehensive guidelines on creating a project plan and using AI technologies to outline and execute it?
"""
```

```python
messages = [
    {"role": "system", "content": "You are a senior project manager and have experience with a variety of IT projects. With the advent of AI, you use AI technologies to execute your project management activities."},
    {"role": "user", "content": question}
    ]
```

```python
# Get gemini to answer, with streaming
from IPython.display import Markdown, display, update_display

stream = await gemini.chat.completions.create(
    model=MODEL_GEMINI, 
    messages=messages,
    stream=True
)

response = ""
display_handle = display(Markdown(""), display_id=True)

async for chunk in stream:
    response += chunk.choices[0].delta.content or ""
    update_display(Markdown(response), display_id=display_handle.display_id)
```

### Via Calling a local LLM Model

```python
# here is the question
question = """
    I am a former software developer that transitioned to a project management role.
    I have been asked to manage a project execution and need to use AI technologies to track the project's progress and performance.
    I am used to using traditional project management tools and techniques, but I am not familiar with how to leverage AI for this purpose.
    Could you please point me to resources or guidance on using AI for project management?
"""
```

```python
messages = [
    {"role": "system", "content": "You are a senior project manager and have experience with a variety of IT projects. With the advent of AI, you use AI technologies to execute your project management activities."},
    {"role": "user", "content": question}
    ]
```

```python
# Get gemma 4 to answer
from openai import OpenAI

openai = OpenAI(base_url="http://127.0.0.1:1234/v1/", api_key="lm-studio")

response = openai.chat.completions.create(
    model=MODEL_GEMA, 
    messages=messages
)
```

```python
display(Markdown(response.choices[0].message.content))
```
